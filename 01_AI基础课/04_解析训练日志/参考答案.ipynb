## 3. 迁移：早停耐心值
写 patience=2 的扫描函数：验证损失连续两轮未刷新最低值时停止。此样例应在第 6 轮触发。

```python
lowest, stale, stop_epoch = float('inf'), 0, None
for row in parsed:
    if row['val'] < lowest:
        lowest, stale = row['val'], 0
    else:
        stale += 1
    if stale >= 2:
        stop_epoch = row['epoch']; break
assert stop_epoch == 6
print('触发早停轮次', stop_epoch)
```
