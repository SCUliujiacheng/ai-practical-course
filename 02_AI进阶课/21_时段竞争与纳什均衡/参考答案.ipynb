```python
coin=np.array([[1.,-1.],[-1.,1.]])
assert pure_equilibria(coin,-coin)==[]
assert check_equilibrium(coin,-coin,[.5,.5],[.5,.5])
es=list(nash.Game(coin,-coin).support_enumeration())
assert any(check_equilibrium(coin,-coin,x,y) for x,y in es)
print(es)
```
